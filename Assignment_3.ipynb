```python
import random

# 1. Setup
secret_number = random.randint(1, 10)
attempts = 3

print("Welcome to Number Guessing Game! You have 3 attempts.")

# 2 & 3. Guessing Logic
while attempts > 0:
    try:
        guess = int(input("Guess the number (between 1 and 10): "))
    except ValueError:
        print("Please enter a valid number.")
        continue

    # Out of range check
    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue # Skip rest of loop

    # Check guess
    if guess > secret_number:
        print("Too high. Try again.")
    elif guess < secret_number:
        print("Too low. Try again.")
    else:
        print("Congratulations! You guessed the correct number.")
        break # Correct guess - exit loop

    attempts -= 1
    print(f"Attempts left: {attempts}")
else:
    # This runs only if while loop completes without break
    print(f"Better luck next time! The number was {secret_number}")
```

```python
# 1. Prompt user
num = int(input("Enter the number for which you want the multiplication table: "))

# 2 & 3. Generate and Display
print(f"\nMultiplication table for {num}:")
for i in range(1, 11): # 1 to 10
    result = num * i
    print(f"{num} x {i} = {result}")
```

```python
# 1. Define function - Type 4: With Arg With Return
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi

# 2. Prompt user
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

# 3. Use function
bmi_value = calculate_bmi(weight, height)

print(f"Your BMI is: {bmi_value:.2f}")

# Optional: Add category
if bmi_value < 18.5:
    print("Category: Underweight")
elif 18.5 <= bmi_value < 24.9:
    print("Category: Normal weight")
elif 25 <= bmi_value < 29.9:
    print("Category: Overweight")
else:
    print("Category: Obesity")
```
